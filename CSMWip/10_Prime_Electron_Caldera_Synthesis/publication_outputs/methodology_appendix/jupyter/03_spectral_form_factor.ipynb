# Spectral Form Factor (SFF) - Caldera Section 7 & 12

Dip-ramp-plateau from prime gap Hamiltonian.

```python
import numpy as np

# Prime gap Hamiltonian: H = ħ/κ Σ g_n^{-1}
def gap_hamiltonian_eigenvalues(gaps, hbar=1, kappa=1):
    """Eigenvalues from prime gaps."""
    return hbar/kappa * np.array([1/g for g in gaps])

# SFF: g(β,t) = |Tr exp(-βH - itH)|^2
def spectral_form_factor(eigenvals, beta, t):
    trace = np.sum(np.exp(-beta * eigenvals - 1j * t * eigenvals))
    return np.abs(trace)**2

# Test with first 1000 prime gaps
gaps = [2, 4, 2, 4, 6, 2, 6, 4, 2, 4]  # sample
eigenvals = gap_hamiltonian_eigenvalues(gaps)
print(f"Eigenvalues: {eigenvals[:5]}...")
```
